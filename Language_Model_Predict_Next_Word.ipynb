```python
# import the libraries

import nltk
from nltk.corpus import reuters
from nltk import trigrams
from collections import  defaultdict
```

```python
# Download the reuters and punk corpus
nltk.download('reuters')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python

# creat a placeholder for the model

model = defaultdict(lambda : defaultdict(lambda: 0))

```

```python
for sentince in reuters.sents():
     for w1, w2, w3 in trigrams(sentince, pad_right = True, pad_left = True):
        model[(w1,w2)][w3] += 1

```

```python
# lets transform the counts to probabilities

for w1_w2 in model :
    total_count = float(sum(model[w1_w2].values()))

    for w3 in model[w1_w2]:
        model [w1_w2][w3] /= total_count


```
